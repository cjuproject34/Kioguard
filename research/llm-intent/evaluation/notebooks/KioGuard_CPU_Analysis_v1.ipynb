# KioGuard CPU 분석 v1

GPU 없이 데이터·라우팅·채점·기존 결과를 분석합니다. 원본 JSONL과 기존 실험 디렉터리는 수정하지 않습니다.


```python
from google.colab import drive
drive.mount("/content/drive", force_remount=False)
REPO_URL = "https://github.com/cjuproject34/Kioguard.git"
REPO_DIR = "/content/Kioguard"
RESULT_ROOT = "/content/drive/MyDrive/KioGuard/experiments/FP16_COMMON_V3"
CPU_OUT = RESULT_ROOT + "/CPU_ANALYSIS_V1"
```

```python
import os, subprocess
if not os.path.exists(os.path.join(REPO_DIR, ".git")):
    subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
else:
    subprocess.run(["git", "-C", REPO_DIR, "pull", "--ff-only"], check=True)
EVAL_ROOT = os.path.join(REPO_DIR, "research", "llm-intent", "evaluation")
print(EVAL_ROOT)
```

```python
import sys
subprocess.run([sys.executable, os.path.join(EVAL_ROOT,"scripts","validate_dataset.py"),
    os.path.join(EVAL_ROOT,"data","selection_dev_v0.1.jsonl"),
    os.path.join(EVAL_ROOT,"data","selection_challenge_v0.1.jsonl"),
    os.path.join(EVAL_ROOT,"data","legacy48_regression_v1.2.jsonl")], check=True)
subprocess.run([sys.executable, os.path.join(EVAL_ROOT,"scripts","audit_evaluation.py"),
    "--core",os.path.join(EVAL_ROOT,"data","selection_dev_v0.1.jsonl"),
    "--challenge",os.path.join(EVAL_ROOT,"data","selection_challenge_v0.1.jsonl"),
    "--out-dir",os.path.join(CPU_OUT,"dataset_audit")], check=True)
```

```python
subprocess.run([sys.executable, os.path.join(EVAL_ROOT,"scripts","analyze_existing_results.py"),
    "--result-root",RESULT_ROOT,"--out-dir",CPU_OUT], check=True)
print("CPU outputs:", CPU_OUT)
```
